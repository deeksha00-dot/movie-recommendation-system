## data understanding

```python

import pandas as pd
import numpy as np
import ast
```

```python
movies = pd.read_csv("../movies_metadata.csv", low_memory=False)

credits = pd.read_csv("../credits.csv")

keywords = pd.read_csv("../keywords.csv")

ratings = pd.read_csv("../ratings_small.csv")
```

```python
movies.sample(5)
```

```python
movies.info()
```

```python
movies.isnull().sum()
```

```python
movies = movies[
    [
        "id",
        "title",
        "overview",
        "genres",
        "release_date",
        "vote_average",
        "vote_count",
        "popularity"
    ]
]
```

```python
movies = movies.dropna(subset=["overview"])
```

```python
movies = movies.fillna("")
```

```python
movies = movies[pd.to_numeric(movies["id"], errors="coerce").notnull()]

movies["id"] = movies["id"].astype(int)
```

```python
credits.head()
```

```python
credits["id"] = credits["id"].astype(int)
```

```python
keywords["id"] = keywords["id"].astype(int)
```

```python
movies = movies.merge(credits, on="id")

movies = movies.merge(keywords, on="id")
```

```python
movies["genres"] = movies["genres"].apply(ast.literal_eval)

movies["cast"] = movies["cast"].apply(ast.literal_eval)

movies["crew"] = movies["crew"].apply(ast.literal_eval)

movies["keywords"] = movies["keywords"].apply(ast.literal_eval)
```

```python
movies.head()
```

```python
def convert(obj):
    L = []

    for i in obj:
        L.append(i['name'])

    return L
```

```python
movies['genres'] = movies['genres'].apply(convert)
```

```python
movies.head()
```

```python
movies['keywords'] = movies['keywords'].apply(convert)
```

```python
def convert_cast(obj):

    L = []

    counter = 0

    for i in obj:

        if counter != 3:

            L.append(i['name'])

            counter += 1

        else:

            break

    return L


movies['cast'] = movies['cast'].apply(convert_cast)
```

```python
movies.head()
```

```python
def fetch_director(obj):

    L = []

    for i in obj:

        if i['job'] == 'Director':

            L.append(i['name'])

            break

    return L
```

```python
movies['crew'] = movies['crew'].apply(fetch_director)
```

```python
movies[['genres','cast','crew','keywords']].head()
```

```python
movies['overview'] = movies['overview'].apply(lambda x: x.split())
```

```python
movies['genres'] = movies['genres'].apply(lambda x: [i.replace(" ", "") for i in x])

movies['cast'] = movies['cast'].apply(lambda x: [i.replace(" ", "") for i in x])

movies['crew'] = movies['crew'].apply(lambda x: [i.replace(" ", "") for i in x])

movies['keywords'] = movies['keywords'].apply(lambda x: [i.replace(" ", "") for i in x])
```

```python
movies['tags'] = (
    movies['overview']
    + movies['genres']
    + movies['keywords']
    + movies['cast']
    + movies['crew']
)
```

```python
movies['tags'] = movies['tags'].apply(lambda x: " ".join(x))
```

```python
movies = movies[['id', 'title', 'tags']]
```

```python
movies.drop_duplicates(inplace=True)
```

```python
movies.to_csv("processed_movies.csv", index=False)
```

```python
metadata = pd.read_csv(
    "../movies_metadata.csv",
    low_memory=False
)
```

```python
movie_details = metadata[
    [
        "id",
        "title",
        "overview",
        "genres",
        "release_date",
        "vote_average",
        "vote_count",
        "poster_path"
    ]
].copy()
```

```python
print(movie_details.shape)
print(movie_details.head())
```

```python
movie_details["overview"] = movie_details["overview"].fillna("")

movie_details["release_date"] = movie_details["release_date"].fillna("")

movie_details["vote_average"] = pd.to_numeric(
    movie_details["vote_average"],
    errors="coerce"
).fillna(0)

movie_details["vote_count"] = pd.to_numeric(
    movie_details["vote_count"],
    errors="coerce"
).fillna(0)

movie_details["poster_path"] = movie_details["poster_path"].fillna("")
```

```python
movie_details["id"] = pd.to_numeric(
    movie_details["id"],
    errors="coerce"
)
```

```python
movie_details = movie_details.dropna(subset=["id"])
```

```python
movie_details["id"] = movie_details["id"].astype(int)
```

```python
movie_details = movie_details.drop_duplicates(
    subset=["id"]
)
```
