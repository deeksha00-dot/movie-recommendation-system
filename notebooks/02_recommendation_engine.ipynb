```python
from sklearn.metrics.pairwise import cosine_similarity

def recommend(movie):
    
    movie_data = movies[
        movies["title"].str.lower() == movie.lower()
    ]

    if movie_data.empty:
        print("Movie not found!")
        return

    movie_index = movie_data.index[0]

    # Calculate similarity only for this movie
    distances = cosine_similarity(
        vectors[movie_index],
        vectors
    ).flatten()

    movie_list = list(enumerate(distances))

    movie_list = sorted(
        movie_list,
        key=lambda x: x[1],
        reverse=True
    )[1:6]

    print(f"\nRecommendations for '{movie}':\n")

    for i, score in movie_list:
        print(movies.iloc[i]["title"])
```

```python
def recommend(movie):
    movie = movie.lower()

    matches = movies[
        movies["title"].str.lower() == movie
    ]

    if matches.empty:
        return []

    movie_index = matches.index[0]

    recommended_indices = recommendations[movie_index]

    return movies.iloc[recommended_indices]["title"].tolist()
```
